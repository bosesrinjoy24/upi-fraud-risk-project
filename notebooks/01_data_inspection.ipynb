# Data Inspection & Quality

This notebook reads the existing project dataset and displays
validation results. It does not download, modify, or overwrite files.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.data_loader import load_data
from src.validation import validate_transactions

df = load_data()

print("Dataset shape:", df.shape)
print("\nColumns:")
print(list(df.columns))

print("\nData types:")
display(df.dtypes.rename("dtype").to_frame())

print("\nMissing values:")
display(df.isna().sum().rename("missing_count").to_frame())

print("\nFraud distribution:")
display(
    df["fraud"].value_counts(dropna=False)
    .rename_axis("fraud")
    .to_frame("count")
)

print("\nFraud rate:", round(float(df["fraud"].mean()), 6))

print("\nFirst five rows:")
display(df.head())

validate_transactions(df)
print("\n[PASS] Dataset validation completed.")
```
