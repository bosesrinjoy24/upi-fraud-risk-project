```python
import os
import io
import contextlib
import pandas as pd

# --- Auto-install Pillow if missing ---
try:
    from PIL import Image, ImageDraw, ImageFont
except ImportError:
    import subprocess, sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "pillow"])
    from PIL import Image, ImageDraw, ImageFont

# --- Load the CSV from the project folder (already copied) ---
csv_path = r"C:\Users\boses\Desktop\bfsi-fraud-project\data\UPI_FRAUD_100K.csv"
df = pd.read_csv(csv_path)

# Make pandas print wide output without wrapping mid-table
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 220)
pd.set_option('display.max_colwidth', 30)

# --- Capture the inspection output as text ---
buf = io.StringIO()
with contextlib.redirect_stdout(buf):
    print("SHAPE:", df.shape)
    print("\nCOLUMNS:", df.columns.tolist())
    print("\nDTYPES:\n", df.dtypes)
    print("\nMISSING VALUES:\n", df.isnull().sum())
    print("\nFIRST 3 ROWS:\n", df.head(3))
    label_col = 'fraud'
    print(f"\nFraud distribution for '{label_col}':")
    print(df[label_col].value_counts())
    print("Fraud rate:", df[label_col].mean())

output_text = buf.getvalue()
print(output_text)  # also show it in the cell for confirmation

# --- Render the text to a PNG ---
lines = output_text.split("\n")

# Font — try Consolas, then Courier New, then Lucida Console, then default
font_candidates = [
    r"C:\Windows\Fonts\consola.ttf",
    r"C:\Windows\Fonts\cour.ttf",
    r"C:\Windows\Fonts\lucon.ttf",
]
font = None
for fp in font_candidates:
    if os.path.exists(fp):
        try:
            font = ImageFont.truetype(fp, 13)
            break
        except Exception:
            continue
if font is None:
    font = ImageFont.load_default()

# Measure char width using the chosen font
tmp = Image.new("RGB", (10, 10))
tmp_draw = ImageDraw.Draw(tmp)
bbox = tmp_draw.textbbox((0, 0), "M" * 20, font=font)
char_width = (bbox[2] - bbox[0]) / 20

line_height = 18
padding = 24
max_line_len = max(len(l) for l in lines) if lines else 1

width = int(max_line_len * char_width) + padding * 2
height = len(lines) * line_height + padding * 2

img = Image.new("RGB", (width, height), color=(30, 30, 30))
draw = ImageDraw.Draw(img)

y = padding
for line in lines:
    draw.text((padding, y), line, fill=(220, 220, 220), font=font)
    y += line_height

# --- Save ---
out_path = r"C:\Users\boses\Desktop\bfsi-fraud-project\outputs\01_data_inspection.png"
os.makedirs(os.path.dirname(out_path), exist_ok=True)
img.save(out_path)

print(f"\nSaved: {out_path}")
print(f"Image size: {img.size[0]} x {img.size[1]} px")
print(f"File size: {round(os.path.getsize(out_path)/1024, 1)} KB")
```

```python
import os, io, contextlib
import pandas as pd
from PIL import Image, ImageDraw, ImageFont

# --- Load data ---
csv_path = r"C:\Users\boses\Desktop\bfsi-fraud-project\data\UPI_FRAUD_100K.csv"
df = pd.read_csv(csv_path)

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)
pd.set_option('display.max_colwidth', 25)

# --- Capture output ---
buf = io.StringIO()
with contextlib.redirect_stdout(buf):
    print("SHAPE:", df.shape)
    print("\nCOLUMNS:", df.columns.tolist())
    print("\nDTYPES:\n", df.dtypes)
    print("\nMISSING VALUES:\n", df.isnull().sum())
    print("\nFIRST 3 ROWS:\n", df.head(3))
    print("\nFraud distribution for 'fraud':")
    print(df['fraud'].value_counts())
    print("Fraud rate:", df['fraud'].mean())

raw = buf.getvalue()
print(raw)

# --- Wrap long lines at 115 chars ---
import textwrap
wrapped_lines = []
for line in raw.split("\n"):
    if len(line) <= 115:
        wrapped_lines.append(line)
    else:
        wrapped_lines.extend(textwrap.wrap(line, width=115,
                                           subsequent_indent="    ",
                                           break_long_words=True,
                                           break_on_hyphens=False))

# --- Render ---
font_path = r"C:\Windows\Fonts\consola.ttf"
font = ImageFont.truetype(font_path, 17) if os.path.exists(font_path) else ImageFont.load_default()

tmp = Image.new("RGB", (10, 10))
bbox = ImageDraw.Draw(tmp).textbbox((0, 0), "M" * 20, font=font)
char_w = (bbox[2] - bbox[0]) / 20

line_h = 23
pad = 28
max_len = max(len(l) for l in wrapped_lines)

width  = int(max_len * char_w) + pad * 2
height = len(wrapped_lines) * line_h + pad * 2

img = Image.new("RGB", (width, height), color=(30, 30, 30))
draw = ImageDraw.Draw(img)

y = pad
for line in wrapped_lines:
    draw.text((pad, y), line, fill=(225, 225, 225), font=font)
    y += line_h

out_path = r"C:\Users\boses\Desktop\bfsi-fraud-project\outputs\01_data_inspection.png"
img.save(out_path)

print(f"\nSaved: {out_path}")
print(f"Image size: {img.size[0]} x {img.size[1]} px")
print(f"File size: {round(os.path.getsize(out_path)/1024, 1)} KB")
```
